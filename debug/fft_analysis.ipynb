```python
import numpy as np
from scipy.fft import rfft, rfftfreq
import matplotlib.pyplot as plt

arr = np.load("udp_capture/component_1.npy")
waveform = arr[:, 1:].flatten()
waveform = waveform[~np.isnan(waveform)]

sample_rate = 48000  # whatever your actual rate is
freqs = rfftfreq(len(waveform), d=1/sample_rate)
spectrum = np.abs(rfft(waveform))

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 6))
ax1.plot(waveform)
ax1.set_title("Waveform")
ax2.plot(freqs, spectrum)
ax2.set_title("FFT Magnitude")
plt.tight_layout()
plt.show()
```
